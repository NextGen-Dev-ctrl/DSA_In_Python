```python
# Stack:
#       The stack is a data structure where every new data is insert in the up of existing data and there is a specific size of the stack
# Stack overflow:
#               When the stack is full and we force to insert the value in it so it is called stack overflow
# Stack underflow:
#               When the stack is empty so we can call it stack underflow
# Types:
#       There are two types of the Stack 
#  1. LIFO (Last in first out)      2. FIFO (first in first out)
# LIFO:
#      In LIFO we cannot went out the thing that is inserted between the stack but we can only get the last thing we can inserted
```

```python
# So the solution of this is we can close the list of the python in the box and it cannot do this type operation
class Stack:
    def __init__(self):
        self.s = []

    def __str__(self):
        ret_str = '['
        for i in self.s:
            ret_str += str(i) + ','
        ret_str = ret_str.rstrip(',')
        ret_str += ']'
        return ret_str

    def push(self,val):
        self.s.append(val)

    def pop(self):
        return self.s.pop()

    def peek(self):
        return self.s[-1]
```

```python
s = Stack()
s.push(1)
s.push(2)
s.push(3)
print(s)
# print(s[2])          <- Now this can give us an error
print(s.pop())
print(s)
print(s.peek())
```

```python
# The stack is only this but there is loot of application of it we can there study the some cause that is useable at this level
# Examples:
# So the first example we can study is Bracket Matching 
# Bracket Matching:
#                  In this the we can use there type bracket ([{}]) and correct this bracket squance using the stack 
# We Know that the came first can be closed at the last like [(]) This is incorrect so the priority of the bracket is checking by the
# stack LIFO structure
# First we can look up the zip structure
opening = '([{'
closing = ')]}'
mapping = dict(zip(opening,closing))
print(mapping)
print(mapping.keys())
print(mapping.values())
```

```python
def is_matched(string):
    opening = '([{'
    closing = ')]}'
    mapping = dict(zip(opening,closing))

    stack = []

    for char in string:
        # case 1 (The character is not a opening and closing braces but some other char like number or math signs)
        if char not in mapping.values() and char not in mapping.keys():
            continue
        # case no 2 
        if char in mapping:  # automatically check the opening braces in the dictionary
            stack.append(mapping[char])  # The match bracket closing brace is stored in the stack
        # case no 3 (check the closing braces is match with the opening brace or not and also there is stack is empty or not)
        elif len(stack) == 0 or char != stack.pop():
            return False
    return len(stack) == 0  # if it return true mean the matching is correct if it can false mean the matching is not correct
```

```python
is_matched("[{(}]")
```

```python
# The next case study of the stack is finding the binary number of the decimal number
# So we have number that we give to it and it can return to the binery
def dec_to_bin(num):
    stack = []
    if num == 0:        # Edge case
        return '0'
    while num != 0:
        remainder = num % 2     # give the remainder that we collected in the stack
        num = num // 2          # it can give the quotiont that we can next number that we can find the remainder
        stack.append(remainder)

    ret = ''
    while stack:
        ret += str(stack.pop())     # come out in the reverse order that we actually want that is we print in the down to up manner

    return ret
```
