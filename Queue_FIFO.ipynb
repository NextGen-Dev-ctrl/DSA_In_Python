```python
# Queue 
#       It is the data structure that can be fixed size and follow the fist in first out structure 
# There are two operation in the Queue
# 1. Enqueue    2. Dequeue
# if the queue is full so we can go the start of the queue because it can follow the circular structure
```

```python
class Queue:
    def __init__(self):
        self.size = 5
        self.queue = list(range(self.size))     # SOME dommy values
        self.i = 0
        self.o = 0

        self.is_empty = True
        self.is_full = False
```

```python
def _inc(self,index):
    if index + 1 == self.size:
        return 0
    else:
        return index + 1
Queue._inc = _inc
```

```python
def enqueue(self, val):
    if self.is_full:
        raise IndexError("The Queue is full. We cannot enqueue")

    self.queue[self.i] = val        # This can add the value in the queue 
    self.i = self._inc(self.i)      # This can increment the position of the i after adding value in previous location

    if self.i == self.o:        # after increment if the i is equal to o means the queue is full
        self.is_full = True

    self.is_empty = False
Queue.enqueue = enqueue
```

```python
def dequeue(self):
    if self.is_empty:
        raise IndexError("The Queue is empty. We cannot dequeue in it")

    ret = self.queue[self.o]        # The can get the value of the current position mean the value that fist come
    self.o = self._inc(self.o)      # This can increment the output variable to the next position after dequeue

    if self.o == self.i:        # if the o is equal to the i after dequeue mean the queue is empty
        self.is_empty = True

    self.is_full = False
    return ret
Queue.dequeue = dequeue
```

```python
def __str__(self):
    return str(self.queue) + ', In: ' + str(self.i) + ', Out: ' + str(self.o)
Queue.__str__= __str__
```

```python
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
# q.enqueue(60) this can give index error

print(q)

print(q.dequeue())
print(q)
print(q.dequeue())
print(q)
print(q.dequeue())
print(q)
print(q.dequeue())
print(q)
print(q.dequeue())
print(q)
# print(q.dequeue())    This can give us an Index error
# print(q)
```
