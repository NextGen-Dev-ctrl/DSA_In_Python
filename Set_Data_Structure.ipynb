```python
# Case study of the set data structure and use of All and Any   
# all()
#       We can give the set of condition in all and it can act like the (and) between the conditions
# any()
#       We can give the set of condition in any and it can act like the (or) between the conditions
# like
a = 5
b = 4
if a > 2 and a < 10 and b > 0 and b < 10:
    print("All constraint stisfied")
```

```python
a = 5
b = 4
constraints = {a > 2 and a < 10 and b > 0 and b < 10}
```

```python
if all(constraints):
    print("All constraint satisfied")
if any(constraints):
    print("At least one constraint satisfied")
```

```python
a = 15
b = 4
constraints = {a > 2 and a < 10 and b > 0 and b < 10}
if all(constraints):
    print("All constraint satisfied")
if any(constraints):
    print("At least one constraint satisfied")
```

```python
# Quantifiers
X = {10,20,30,40,50}
# ulta A xEX,x<100
a = all([x<100 for x in X]) # This can itrate the value of capital x and check the conditaion x<100 and return the list of bool values and 
                        # all can evaluate that if it can all ture so give true other than it can give false
# ulta E xEX,x<100
b = any([x<100 for x in X])  # This can itrate the value of capital x and check the conditaion x<100 and return the list of bool values and 
                        # any can evaluate that if it can any of them is ture so give true other than it can give false
print(a)
print(b)
```

```python
def is_div_20(number):
    return number % 20 == 0
```

```python
all([is_div_20(x) for x in X])  # We can also set the function at the place of condition that can give the bool return
```
