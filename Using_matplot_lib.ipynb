```python
def fib1(i):
    if i<=1:
        return 1
    else:
        return fib1(i-2) + fib1(i-1)
```

```python
def fib2(n):
    a = 1
    b = 1 
    for i in range(n):
        a,b = b,a+b

    return a
```

```python
print("fib1 :")
for i in range(15):
    print(fib1(i), end=",")

print("\n\n")
print("fib2 :")
for j in range(15):
    print(fib2(j),end=",")
```

```python
import time

def compute_time(function, limits): # for the computing time taken we take function and its limit like how many range of value we given to it
    l= []

    for i in range(limits):     # find the time of every call
        start_time = int(round(time.time() * 1000)) # start time check like stop watch start

        function(i) # call the function

        end_time = int(round(time.time() * 1000)) # end time check like stop watch stop

        time_taken = end_time - start_time # calculte the time by minus the end and start time

        l.append(time_taken) # collect all time results in the list

    return l
```

```python
limit = 30
fib1_time = compute_time(fib1,limit)
fib2_time = compute_time(fib2,limit)
```

```python
print("fib1 time taken :")
for i in fib1_time:
    print(i,end=",")

print("\n\n")

print("fib2 time taken :")

for i in fib2_time:
    print(i,end=",")
```
