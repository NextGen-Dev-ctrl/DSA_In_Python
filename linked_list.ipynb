```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
```

```python
# Push operation ha two cases 
# 1.The head none and we insert the first node
# 2.At least one node is present
def Push(self,val):
    new_node = Node(val)

    # First case
    if self.head is None:
        self.head = new_node
        return

    # Second case
    last = self.head
    while last.next is not None:
        last = last.next

    last.next = new_node

LinkedList.Push = Push
```

```python
def __str__(self):
    ret_str = "["
    temp = self.head
    while temp is not None:
        ret_str += str(temp.val) + ", "
        temp = temp.next

    ret_str = ret_str.rstrip(", ")
    ret_str += "]"
    return ret_str
LinkedList.__str__ = __str__
```

```python
l = LinkedList()
l.Push(3) # first case execute
l.Push(4) # second case execute 
l.Push(6) # second case execute
print(l)
```

```python
# POP operation ha two cases\
# 1.Head is pointing only one node
# 2.At least two nodes are present
def Pop(self):
    if self.head is None:
        raise Exception("Cannot PoP. No value")

    # case no 1
    if self.head.next is None:
        val = self.head.val
        self.head = None        # pointing the head to the none so the node is deleted
        return val
    
    # case no 2

    temp = self.head
    while temp.next is not None:
        pre = temp              # the temp is pointing the last node 
        temp = temp.next        # the pre is pointing the second last node
    val = temp.val
    pre.next = None             # the second last node is pointing to nothing So last node is automatically delete
    return val
LinkedList.Pop = Pop
```

```python
print(l)
l.Pop()
print(l)
```

```python
# INSERT operation contain two cases:
# 1. We insert at the 0 index mean at the begining of the list 
# 2. We insert between any two nodes
def Insert(self,val,index = None):
    new_node = Node(val)
    #if the person cannot send the index
    if index is None:
        raise Exception("You cannot send any index")
    # case no 1
    if index == 0:
        new_node.next = self.head
        self.head = new_node
        return
    # case no 2
    temp = self.head
    counter = 0
    while temp is not None and counter < index: # it can also handle the thing like adding at index that is less than the size of the list
        prev = temp                             # it can also handle to add node at the last of the list
        temp = temp.next
        counter += 1
    
    prev.next = new_node
    new_node.next = temp

LinkedList.Insert = Insert 
```

```python
l.Insert(24,0)
print(l)

l.Insert(7,11)
print(l)
```

```python
# REMOVE operation 
# 1. If first node is present and same as value
# 2. Find the value and in complete list and remove this
def Remove(self,val):
    temp = self.head

    # case no 1
    if temp is not None:
        if temp.val == val:
            self.head = temp.next
            temp = None
            return
    
    while temp is not None:
        if temp.val == val:
            break
        pre = temp
        temp = temp.next

    if temp is None:
        return
    pre.next = temp.next

LinkedList.Remove = Remove
```

```python
l.Remove(3)
print(l)
```

```python
# Length of the list 
def len(self):
    temp = self.head
    count = 0
    while temp is not None:
        count += 1
        temp = temp.next
    
    return count

LinkedList.len = len
```

```python
# Get() function to get value using the index
# case 1: the list is empty 
# case 2: the value againt the index
# case 3: user can give the wrong index value we can handle it 
def get(self,index):

    temp = self.head 
    # case no 1
    if temp is None:
        raise Exception("The list is empty")
    
    # case no 3
    counter = 0 
    while temp is not None:
        counter += 1
        temp = temp.next
    
    # check weather the given index is in range than return value
    # case no 2
    if index <= counter:
        temp = self.head
        counter2 = 0
        while counter2 != index:
            counter2 += 1
            temp = temp.next
        return temp.val
    else:
        raise Exception("Index is not in range")
    
LinkedList.get = get
```
