```python
# Recursion:
#           The standerd def of recursion that extremely use is "The function that calling itself" but the originally the recursion is the
# function that calling another function and so on and at least the again the first function is call at the end
# Direct Recursion:
#                   The function calling itself again and again is called direct recursion
# Indirect Recursion:
#                   The function calling another function and at the end it can call itself is called indirect recursion
# We know the recursion so now we discuss some important example of it but the most commenly example come in the advance data structure
# There are two cases in the Recursion
# Base Case:
#           The case to stop the recursion if this cannot exist the recursion cannot stop
# Indection Case:
#               The case which is used to recusive the function to n number of times
```

```python
# First Example of the Recursion is Fictorial
def fic(n):
    # Base case
    if n <=1:
        return 1
    # Indection case
    return n * fic(n - 1)           # the value that return can multiply the last number
```

```python
for i in range(1,7):
    print(f"Factorial of {i} is {fic(i)}")
```

```python
# Second example of a recursion is a fibonachi number finding
# fibonachi with recursion
def fib(number):
    if number <= 1:
        return 1

    return fib(number - 1) + fib(number - 2)
```

```python
%timeit fib(20)         # Time to find fib of 20
```

```python
# fibonachi without recursion
def fib1(number):
    if number <= 1:
        return 1 
    else:
        a,b = 1,1
        for i in range(number):
            a,b = b,a+b
        return a
```

```python
%timeit fib1(20)
```

```python
# The algorithum that is fast and also use recursion 
# but this is difficult to understand this
def fib2(number,a = 1, b = 1):
    if number == 0:
        return a
    return fib2(number - 1, b , a+b)  # When everything done at the last call is called "Tail recursion" they can return thing dirctly to the first caller
```

```python
%timeit fib2(20)
```

<h1 style= "font-weight: bold;">Functional Programming</h1>
<p>The power of functional programming to solve problems through the function we use recursion to do task that we can do with itration</p>

```python
def sum_list_recursive(list):
    if len(list) == 0:      # base case if the length of the list is 0 we return 0
        return 0

    return list[0] + sum_list_recursive(list[1:])       # Indective case: Head + sum of the other values
```

```python
l = [2,2,2]
sum_list_recursive(l)
```
