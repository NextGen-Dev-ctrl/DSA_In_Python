```python
class point:

    def __init__(self,x=0,y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return "[" + str(self.x) + "," + str(self.y) + "]"
```

```python
p1 = point()
print("p1 ",p1.x)

p2 = point(2,4)
print("p2 ",p2.x)

print(p2)
```

```python
class Shape:
    def __init__(self,points):
        self.points = points

    def __str__(self):
        ret = ""

        for i in self.points:
            ret += str(i) + " - "

        return ret
t1 = point(5,5)
t2 = point(10,5)
t3 = point(5,10)
p = [t1,t2,t3]

sh = Shape(p)
```

```python
print(sh)
```

```python
def print_points(self):  #this is only the simple function
    for i in self.points:
        print(i)

Shape.print_points = print_points   #now this function is also added in the class shape

sh.print_points() #now with the referance variabel of the class shape we can access this print_points
```
