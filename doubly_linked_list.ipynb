```python
# doubly linked list is same as the linked list but one difference is:
# Linked List:
#        Contain only the address of the next node
#        We can move only one direction of the list from head to end 
#        We cannot go backword at any position of the list 
#        if we can cross the disired node we can again loop whole of the list 
# Doubly linked list:
#         Contain the address of both next node and previous node
#         We can move in both direction of the list both back and forth
#         We can do operation on both back and forward direction
```

```python
class Node:
    def __init__(self,val = None):
        self.val = val
        self.next = None
        self.prev = None

class Doubly_Linked_list:
    def __init__(self):
        self.head = None
```

```python
# Push operation 
# case 1:
#       The head is none set the head to the new node and previous to the None
# case 2:
#       At least one node exist set the last node "next" to the new node and prev to the last node
def push(self,value):

    new_node = Node(value)
    # case 1
    if self.head is None:
        self.head = new_node
        return

    # case 2 (first we can go to the last node)
    temp = self.head
    while temp.next is not None:
        temp = temp.next

    temp.next = new_node
    new_node.prev = temp  # Now the new inserted node contain the address of previous node

Doubly_Linked_list.push = push
```

```python
def __str__(self):
    ret_str = "["
    temp = self.head
    while temp is not None:
        ret_str += str(temp.val) + ", "
        temp = temp.next

    ret_str = ret_str.rstrip(", ")
    ret_str += "]"
    return ret_str
Doubly_Linked_list.__str__ = __str__
```

```python
l = Doubly_Linked_list()
l.push(3)
l.push(4)
l.push(5)
print(l)
```

```python
# POP operation 
# case 1:
#       If contain only one node set the head to the None
# case 2:
#       First go to the last node set the second last node to the None same as linked list there is no difference in it
def pop(self):

    if self.head is None:
        raise Exception("List is empty")

    # case 1
    if self.head.next is None:
        self.head = None
        return
    
    # case 2
    temp = self.head
    while temp.next is not None:
        prev = temp       # the temp is pointing the last node 
        temp = temp.next        # the pre is pointing the second last node

    prev.next = None         # the second last node is pointing to nothing So last node is automatically delete

Doubly_Linked_list.pop = pop
```

```python
l.pop()
print(l)
```

```python
# Inset operation
# case 1:
#       Inset value at 0 index So the head is pointing to the new node and new node prev is pointing to None and the second node prev pointing
#       to the new node
# case 2:
#       Inset value other than 0 index So the prev pointer can pointing to the new node and the new node pointing to the temp next node and 
#       also set the prev one
def insert(self,index,value):

    new_node = Node(value)

    # case 1
    if index == 0:
        new_node.next = self.head
        
        if self.head is not None:       # Handle the case if the self.head is pointing to nothing so it can cannot set the prev to the new
            self.head.prev = new_node   # node which already make no sense because it is pointing to None 

        self.head = new_node

        return
    
    # case 2
    temp = self.head 

    counter = 0 
    while temp is not None and counter < index:  # this loop is doing if the index is larger than the size of the list is can insert the
        prev = temp                              # at the end of the list and also check the position if it is correct
        temp = temp.next
        counter += 1

    prev.next = new_node
    new_node.prev = prev

    new_node.next = temp
    temp.prev = new_node 
```
