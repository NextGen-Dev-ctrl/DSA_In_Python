```python
#Binary Tree:
#       Binary tree is data structe that contain parent node and each parent node contain at most 2 child node it start with the one node
# called 'Root' and the end node is called 'leaf' when we go to the root to the down side we called it level's 
```

```python
class TreeNode:
    def __init__(self,x):
        self.value = x
        self.left = None
        self.right = None
```

```python
def print_tree(tree,level=0,label='.'):
    print(' '*(level*2) + label + ':',tree.value)
    for child, lbl in zip([tree.left, tree.right],['L','R']):
        if child is not None:
            print_tree(child,level+1,lbl)
```

```python
# Left Tree
t1 = TreeNode(2)

t1.left = TreeNode(3)
t1.right = TreeNode(4)
print_tree(t1)
```

```python
t1.left.left = TreeNode(5)
t1.left.right = TreeNode(6)

t1.right.right = TreeNode(7)
t1.right.right.left = TreeNode(8)
t1.right.right.right = TreeNode(9)
t1.right.right.right.right = TreeNode(8)
t1.right.right.right.left = TreeNode(9)

print_tree(t1)
```

# Tree Traversel
<h4>Depth First Search(DFS)</h4>

```python
# print all nodes in the tree 
def DFS(self):              # generally it is called pre-order mean the root process firt then traverse left and right
    print(self.value)

    if self.left:           # if self.left give None it means False
        self.left.DFS()
    if self.right:          # same as there
        self.right.DFS()

TreeNode.DFS = DFS
```

```python
t1.DFS()
```

```python
def in_order_DFS(self):
    if self.left:
        self.left.in_order_DFS()

    print(self.value)

    if self.right:
        self.right.in_order_DFS()

TreeNode.in_order_DFS = in_order_DFS
```

```python
t1.in_order_DFS()
```

```python
def post_order_DFS(self):
    if self.left:
        self.left.post_order_DFS()

    if self.right:
        self.right.post_order_DFS()

    print(self.value)

TreeNode.post_order_DFS = post_order_DFS
```

```python
t1.post_order_DFS()
```

# Tree Traversel
<h4>Breadth First Search(BFS)</h4>

The algorithm of BFS is we process the node and we add it's child in Queue and pop it 

```python
def BFS(self):

    to_visit = [self]

    while to_visit:         # The loop will continue while the list is not empty 
        current = to_visit.pop(0)    # The pop zero can get the first value insted of last

        print(current.value)

        if current.left:
            to_visit.append(current.left)
        if current.right:
            to_visit.append(current.right)

TreeNode.BFS = BFS
```

```python
t1.BFS()
```

# Perform Arbitrary Task on all Nodes

We want to sum all the values of tree so we cannot do because we can use recursive call <br>
The concept we can use there is called a Hock <br>
We call the function from the logic and give the node as a parameter <br>
This function can apply the operation on the node it can be any (add,sub,max) etc 

```python
def DFS_apply(self,function):
    function(self)

    if self.left:
        self.left.DFS_apply(function)
    if self.right:
        self.right.DFS_apply(function)

TreeNode.DFS_apply = DFS_apply 
```

```python
# This class not link with the tree class we can make it for over ease
class Perform_Sum:
    def __init__(self):
        self.sum = 0

    def process(self,node):
        self.sum += node.value

    def get_sum(self):
        return self.sum

    def reset_sum(self):
        self.sum = 0

P = Perform_Sum()
```

```python
t1.DFS_apply(P.process)
print(P.get_sum())
```
