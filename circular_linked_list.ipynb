```python
# Circular Linked List:
#                       A circular linked list is similar as linked list but one differnece it cannot end at the specific point but can
# triverse to the whole list again and again continuesly this can help them to also done the process of the node that we can left due to any
# reason.
# Benefits:
# In the bank the token system can use this to again figure out the customers that can left due to any reason 
```

```python
class Node:
    def __init__(self,val=None):
        self.val = val
        self.next = None

class Circular_linked_list:
    def __init__(self):
        self.head = None
```

```python
def __str__(self):
    ret_str = '['
    temp = self.head
    while temp is not None:         # This can also handle that at least one value exist mean list is not empty
        ret_str += str(temp.val) + ','
        temp = temp.next

        if temp == self.head:       # This can break the loop if one circle is complete also handle it if there is only one node 
            break                   # It can break the loop 

    ret_str = ret_str.rstrip(',')
    ret_str += ']'
    return ret_str

Circular_linked_list.__str__ = __str__
```

```python
# special function that give the last node of the list at any case
def _get_last(self):

    # no node,not last
    if self.head is None:
        return None
    
    # Only one node exist
    if self.head.next == self.head:
        return self.head

    # At least two nodes can exist in the list (This cause can also handle the case no 2 because temp.next == self.head if only one node exist)
    temp = self.head.next 
    while temp.next != self.head:
        temp = temp.next
    
    return temp

Circular_linked_list._get_last = _get_last
```

```python
# Insert Operation
# There are same two causes in it we can insert at 0 index
# and second one we can insert at any point in the list other than the 0 index 
# butt doing this is different in circular linked list
def insert(self,index,value):

    # first we can make a new node
    new_node = Node(value)

    # find the last of the list
    last = self._get_last()

    # First handle the case as insertion at 0 index
    if index == 0:
        new_node.next = self.head
        self.head = new_node
        # if there is only one node so we can set it to again it
        if last is None:
            self.head.next = self.head  # only one node exist and it can also pointing it self
        else:
            last.next = new_node # if more than one node exist set the last one to the new_node that is now the first node
        return

    # this cause can handle the problem that if the list is empty and user can give index other than 0 so this cause can execute
    if self.head is None and index > 0:
        raise IndexError("Cannot insert at " + str(index) + " because list cannot exist")
    
    # case 2 for other indices
    temp = self.head

    counter = 0
    while temp is not None and counter < index: # if the person enter index that is greater than the list the loop is contniously looping
        prev = temp                             # at the end where it can stop it can insert the value because this is a circular linked list 
        temp = temp.next                        # we can insert it at any place
        counter += 1

    prev.next = new_node
    new_node.next = temp

Circular_linked_list.insert = insert
```

```python
l = Circular_linked_list()
l.insert(0,1)
l.insert(1,2)
l.insert(2,3)
print(l)
```

```python
# To check that this thing actually a circular linked list 
l._get_last().next == l.head
```

```python
# remove operation
# Case 1:
#       The list is empty raise Exception
# Case 2:
#       One node exist and it can match to the val 
# Case 3:
#       At least two node exist and we want to remove the first value then we set head to the second value and last to the head
# Case 4:
#       Remove the value between the list 
def remove(self,val):
    # Case 1
    if self.head is None:
        raise Exception("The list is empty")
        
    # Case 2
    temp = self.head
    last = self._get_last()
    if temp.val == val:
        if last == self.head:   # only one node exist and program set it to None after removel
            self.head = None
        # Case 3
        else:
            self.head = temp.next       # if case match and at least two node exist we can set the head to the second node and update the last
            last.next = self.head       # and set it to the head
        return
    # Case 4
    prev = temp
    temp = temp.next
    while temp != last.next:
        if temp.val == val:
            break
        prev = temp
        temp = temp.next

    if temp == last.next:       # if the value is not found simply return
        return

    prev.next = temp.next

Circular_linked_list.remove = remove
```
